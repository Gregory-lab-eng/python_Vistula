```python
# Zmiana tekstu na lowercase + tokenizacja
words = re.findall(r'\w+', sample_text.lower())

# Build Markov chain: word -> list of next words
markov_chain = defaultdict(list)
for current_word, next_word in zip(words[:-1], words[1:]):
    markov_chain[current_word].append(next_word)

# Generacja nastepnego slowa
def predict_next_word(current_word):
    next_words = markov_chain.get(current_word, [])
    if not next_words:
        return None
    return random.choice(next_words)

# generacja sentencji
def generate_text(start_word, length=15):
    current_word = start_word.lower()
    generated = [current_word]

    for _ in range(length):
        next_word = predict_next_word(current_word)
        if not next_word:
            break
        generated.append(next_word)
        current_word = next_word

    return ' '.join(generated)
```

```python

# ================================
# Tokenizacja (bardzo uproszczona)
# ================================
vocab = set()
for sentence in sentences:
    vocab.update(sentence.split())

# Tworzymy słownik {słowo: indeks}
word2idx = {word: idx+2 for idx, word in enumerate(sorted(vocab))}
word2idx["<PAD>"] = 0
word2idx["<SOS>"] = 1
idx2word = {idx: word for word, idx in word2idx.items()}

vocab_size = len(word2idx)

def encode(sentence):
    return [word2idx[word] for word in sentence.split()]

def decode(indices):
    return ' '.join([idx2word[idx] for idx in indices if idx > 1])
```

```python
# ================================
# Funkcja tworząca batch danych
# ================================
def make_batch(sentences):
    encoded = [encode(s) for s in sentences]
    max_len = max(len(s) for s in encoded)
    padded = [s + [0]*(max_len - len(s)) for s in encoded]
    return torch.tensor(padded).T  # shape: [seq_len, batch_size]
```

```python
# ================================
# KODOWANIE POZYCYJNE (Positional Encoding)
# ================================
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.pe = pe.unsqueeze(1)  # shape: [max_len, 1, d_model]

    def forward(self, x):
        return x + self.pe[:x.size(0), :]
```

```python
# ================================
# PROSTY MODEL TRANSFORMERA
# ================================
#Zad 1. Wyjaśnijcie, czym sś parametry wejściowe w __init__: emb_size, nhead, num_layers
class SimpleTransformer(nn.Module):
    def __init__(self, vocab_size, emb_size=64, nhead=4, num_layers=2):
        super(SimpleTransformer, self).__init__()

        # --- WORD EMBEDDING ---
        self.embedding = nn.Embedding(vocab_size, emb_size)

        # --- POSITIONAL ENCODING ---
        self.pos_encoder = PositionalEncoding(emb_size)

        # --- SELF-ATTENTION + RESIDUAL CONNECTIONS ---
        # TransformerEncoderLayer zawiera:
        # - mechanizm samo-uwagi (self-attention)
        # - normalizację warstw (LayerNorm)
        # - połączenia rezydualne (residual connections)
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=emb_size, nhead=nhead),
            num_layers=num_layers
        )

        # --- FULLY CONNECTED LAYER (DEKODER) ---
        self.decoder = nn.Linear(emb_size, vocab_size)

    def forward(self, src):
        embedded = self.embedding(src)            # WORD EMBEDDING
        embedded = self.pos_encoder(embedded)     # POSITIONAL ENCODING
        out = self.transformer(embedded)          # SELF-ATTENTION + RESIDUALS
        return self.decoder(out)                  # FULLY CONNECTED LAYER
```

```python
# ================================
# TRENING MODELU
# ================================
#ZAD 2. Pprawncie błąd - model i dane muszą być na tym samym urządzeniu. W tym przypadku model jest offloadowany na GPU, ale dane są nadal na CPU.
####### Wskazówka - należy poprawić jedna z poprzednich sencji, a nie sekcję Trening modelu!
model = SimpleTransformer(vocab_size).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()

# Przygotowanie danych treningowych
src = make_batch(sentences).to(device)
tgt = src.clone().to(device)

# Trening
epochs = 100
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()
    output = model(src)
    # Use .reshape(...) instead of .view(...) for non-contiguous tensors
    loss = loss_fn(output.reshape(-1, vocab_size), tgt.reshape(-1))
    loss.backward()
    optimizer.step()
    if (epoch + 1) % 10 == 0:
        print(f"Epoka {epoch+1}, Strata: {loss.item():.4f}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math
from collections import defaultdict
```

```python
# ================================
# Konfiguracja urządzenia (CPU lub GPU)
# ================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ================================
# Dane treningowe — sentencje motywacyjne po polsku
# ================================
sentences = [
    "Życie to podróż nie cel",
    "Każdy dzień to nowa szansa",
    "Cisza mówi więcej niż słowa",
    "Nadzieja umiera ostatnia",
    "Miłość to wybór nie przypadek",
    "Niebo jest zawsze nad chmurami",
    "Zmiana zaczyna się od decyzji",
    "Nic nie trwa wiecznie",
    "Najtrudniejsze decyzje są często najlepsze",
    "Czas leczy rany",
    "Szczęście jest stanem umysłu",
    "Nie bój się marzyć",
    "Każda burza kiedyś się kończy",
    "Milczenie też jest odpowiedzią",
    "Nie wszystko złoto co się świeci",
    "Największa siła tkwi w prostocie",
    "Rób to co kochasz",
    "Nie oceniaj książki po okładce",
    "Małe kroki tworzą wielkie zmiany",
    "Znajdź czas dla siebie",
    "Przeszłość nie definiuje przyszłości",
    "Radość kryje się w szczegółach",
    "Lepiej być niż mieć",
    "Zacznij od siebie",
    "Każdy dzień to nowy początek",
]

# ================================
# Tokenizacja — prosty tokenizer
# ================================
vocab = set()
for sentence in sentences:
    vocab.update(sentence.split())

word2idx = {word: idx + 2 for idx, word in enumerate(sorted(vocab))}
word2idx["<PAD>"] = 0
word2idx["<SOS>"] = 1
idx2word = {idx: word for word, idx in word2idx.items()}
vocab_size = len(word2idx)

def encode(sentence):
    return [word2idx[word] for word in sentence.split()]

def decode(indices):
    return ' '.join([idx2word.get(idx, "") for idx in indices if idx > 1])
```

```python
# ZAD 1. Wyjasnijcie czym rozni sie ponizsza funkcja make_batch od poprzedniej, ktora napisalismy na poprzednich cwiczeniach
# ================================
# Tworzenie batchy: wejście i target przesunięte o 1
# ================================
def make_batch(sentences):
    encoded = [encode(s) for s in sentences]
    max_len = max(len(s) for s in encoded)
    padded = [s + [0]*(max_len - len(s)) for s in encoded]
    src = [s[:-1] for s in padded]  # wejście: bez ostatniego tokena
    tgt = [s[1:] for s in padded]   # target: bez pierwszego tokena
    return torch.tensor(src).T, torch.tensor(tgt).T  # [seq_len, batch_size]

#Zad 2. Dodajcie instrukcje do zofflodowania danych to odpowiedniego device to(device)
# ================================
# Kodowanie pozycyjne (dla kolejności słów)
# ================================
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.pe = pe.unsqueeze(1)

    def forward(self, x):
        return x + self.pe[:x.size(0), :].to(x.device)

# ================================
# Prosty model transformera (tylko encoder)
# ================================
class SimpleTransformer(nn.Module):
    def __init__(self, vocab_size, emb_size=64, nhead=4, num_layers=2):
        super().__init__()
        # --- Word Embedding ---
        self.embedding = nn.Embedding(vocab_size, emb_size)

        # --- Positional Encoding ---
        self.pos_encoder = PositionalEncoding(emb_size)

        # --- Self-Attention + Residual Connections ---
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=emb_size, nhead=nhead),
            num_layers=num_layers
        )

        # --- Fully Connected Layer (decoder) ---
        self.decoder = nn.Linear(emb_size, vocab_size)

    def forward(self, src):
        x = self.embedding(src)
        x = self.pos_encoder(x)
        x = self.transformer(x)
        return self.decoder(x)
```

```python
# ================================
# Trening modelu
# ================================
model = SimpleTransformer(vocab_size).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()

src, tgt = make_batch(sentences)
src, tgt = src.to(device), tgt.to(device)

epochs = 10000
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()
    output = model(src)
    loss = loss_fn(output.reshape(-1, vocab_size), tgt.reshape(-1))
    loss.backward()
    optimizer.step()
    if (epoch + 1) % 50 == 0:
        print(f"Epoka {epoch+1}, Strata: {loss.item():.4f}")

# ================================
# Generowanie sekwencji słowo po słowie
# ================================
def generate_sequence(model, start_word, max_len=10):
    model.eval()
    tokens = [word2idx.get(start_word, 1)]  # zaczynamy od jednego słowa

    for _ in range(max_len):
        input_tensor = torch.tensor(tokens).unsqueeze(1).to(device)
        with torch.no_grad():
            output = model(input_tensor)

        next_token_logits = output[-1, 0]
        probabilities = F.softmax(next_token_logits, dim=0)
        next_token = torch.multinomial(probabilities, num_samples=1).item()

        if next_token == 0:
            break
        tokens.append(next_token)

    return decode(tokens)
```

```python
#ZAD 1. Zalozcie konto w firecrawl, stworzcie API key. Nastepnie dodajcie secret o nazwie FIRECRAWL_API_KEY, ktory bedzie zawieral API key
#ZAD 2. Zdefiniujcie temat, na ktory chcecie napisac artykul
%%writefile /content/book_writing_flow/tools/custom_tool.py
import os
import requests
from typing import Type
from pydantic import BaseModel, Field
from crewai.tools import BaseTool
from google.colab import userdata
import os

# Load the API key from Colab secrets and set it as an environment variable
os.environ["FIRECRAWL_API_KEY"] = userdata.get('FIRECRAWL_API_KEY')

class FirecrawlWebSearchToolInput(BaseModel):
    """Input schema for FirecrawlWebSearchTool."""

    title: str = Field(..., description="......................................")


class FirecrawlWebSearchTool(BaseTool):
    name: str = "FirecrawlWebSearchTool"
    description: str = "Tool to retrieve Google results via Firecrawl."
    args_schema: Type[BaseModel] = FirecrawlWebSearchToolInput

    def _run(self, title: str) -> dict:
        url = "https://api.firecrawl.dev/v1/search"
        headers = {
            "Authorization": f"Bearer {os.getenv('FIRECRAWL_API_KEY')}",
            "Content-Type": "application/json",
        }
        payload = {"query": title, "limit": 5, "lang": "pl"}

        response = requests.post(url, json=payload, headers=headers)

        return response.json()
```

```python
#!/usr/bin/env python
import os
import asyncio
from dotenv import load_dotenv
from pydantic import BaseModel
from crewai.flow import Flow, listen, start
from google.colab import userdata
import os
#from book_writing_flow.crews.Writer_crew.writer_crew import ChapterWriterCrew
#from book_writing_flow.crews.Outline_crew.outline_crew import OutlineCrew
#from llama_index.llms.openai import OpenAI
#import os
#from book_writing_flow.tools.custom_tool import FirecrawlWebSearchTool


load_dotenv()


class Chapter(BaseModel):
    title: str = ""
    content: str = ""


class BookState(BaseModel):
    topic: str = "administracja publiczna"
    total_chapters: int = 0
    titles: list[str] = []
    chapters: list[Chapter] = []


class BookFlow(Flow[BookState]):

    @start()
    def generate_outline(self):
        print("Generating outline")
        outline = OutlineCrew().crew().kickoff(inputs={"topic": self.state.topic})
        self.state.total_chapters = outline.pydantic.total_chapters
        self.state.titles = outline.pydantic.titles

    @listen(generate_outline)
    async def generate_chapters(self):
        print("Generating chapters")
        tasks = []

        async def write_single_chapter(title: str):
            result = (
                ChapterWriterCrew()
                .crew()
                .kickoff(
                    inputs={
                        "title": title,
                        "topic": self.state.topic,
                        "chapters": [chapter.title for chapter in self.state.chapters],
                    }
                )
            )
            return result.pydantic

        # Create tasks for each chapter
        for i in range(self.state.total_chapters):
            task = asyncio.create_task(write_single_chapter(self.state.titles[i]))
            tasks.append(task)

        # Wait for all chapters to be generated concurrently
        chapters = await asyncio.gather(*tasks)
        print(f"Generated {len(chapters)} chapters")
        self.state.chapters.extend(chapters)

    @listen(generate_chapters)
    def save_book(self):
        print("Saving book")
        with open("book.md", "w") as f:
            for chapter in self.state.chapters:
                f.write("# " + chapter.title + "\n")
                f.write(chapter.content + "\n")


# Modify the kickoff function to be an async function
async def kickoff():
    book_flow = BookFlow()
    # Directly await the async kickoff in a Jupyter environment
    await book_flow.kickoff_async()


def plot():
    book_flow = BookFlow()
    book_flow.plot()


if __name__ == "__main__":
    # In a Jupyter notebook, you can just await the async function directly.
    # If running as a script, you would use asyncio.run(kickoff())
    try:
        get_ipython().run_line_magic('autoawait', 'True') # Ensure autoawait is enabled in IPython
        await kickoff()
    except NameError: # Handles cases where get_ipython is not available (i.e., running as a standard script)
        asyncio.run(kickoff())
```
